```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
import numpy as np
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([faqs])
```

```python
len(tokenizer.word_index)
```

```python
input_sequences = []
for sentence in faqs.split('\n'):
    tokenized_sen = tokenzier.texts_to_sequences([sentence])[0]
    for i in range(1, len(tokenized_sen)):
        n_gram_sequence = tokenized_sen[:i+1]
        input_sequences.append(n_gram_sequence)
```

```python
input_sequences
```

```python
max_len = max([len(x) for x in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
input_padd_sequences = pad_sequences(input_sequences, maxlen=max_len, padding='pre')
```

```python
X = input_padd_sequences[:,:-1]
y = input_padd_sequences[:,-1]
```

```python
X.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y,num_classes=246)
```

```python
y.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense

model = Sequential()
model.add(Input(shape=(17,)))
model.add(Embedding(246, 100))
model.add(LSTM(150))
model.add(Dense(246, activation='softmax'))

model.summary()
```

```python
model.compile(loss='categorical_crossentropy', optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=300)
```

```python
text = "what is test dataset"

for i in range(10):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]

  # padding (maxlen must match your model's Input layer shape)
  padded_token_text = pad_sequences([token_text], maxlen=17, padding='pre')

  # predict
  pos = np.argmax(model.predict(padded_token_text, verbose=0))

  # Map index to word efficiently
  if pos in tokenizer.index_word:
    word = tokenizer.index_word[pos]
    text = text + " " + word
    print(text)
  else:
    break
```
